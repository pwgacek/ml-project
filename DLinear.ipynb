### Parametry w skrypcie

Dla DLinear `--enc_in` oznacza ilość cech ze zbioru danych, czyli `ilość kolumn - 1` (bo nie bierzemy pod uwagę daty).

Dodatkowo `--individual = False (default)` oznacza, że każda cecha dzieli wspólną warstwę liniową, w paperze oznaczają to jako DLinear-S i jest to domyślne czego używali w eksperymentach.
